# **Running Flyer Analysis Scripts**

```python
import json
from Flyer_Recast_Analysis import run_pipeline 
from Flyer_Recast_Analysis_preFlatten import run_pipeline_flattened 
#run_pipeline and run_pipeline_flattened are essentially the same methods, but run_pipeline_flattened will flatten the surface before detecting the peaks
#while run_pipeline detects peaks on raw surface and then flattens.
```

```python
flyer_names=["F1184_01.xyz"] #change based on what flyer(s) you want to analyze
for flyer_name in flyer_names: 
    #output: dictionary containing ellipse fitting, points, edge stats, and FWHM stats for inner and outer edge
    #change run_pipeline to run_pipeline_flattened if you want to detect peaks after flattening the entire thing first 
    plot_name=flyer_name[:-4]+"_Ridge_Detection.png"
    results = run_pipeline(
        flyer_name,skip_header=14,
        reject_outliers=True,plot=plot_name
    ) 
    #concising results to include a single best estimate(median) for the ridge height and the FWHM
    concised_results={
        'inner': {'median_ridge_height': results["inner"]["stats"]['median'], 'median FWHM': results["inner"]["fwhm_stats"]["median"]},
        'outer': {'median_ridge_height': results["outer"]["stats"]['median'], 'median FWHM': results["outer"]["fwhm_stats"]["median"]}
    }
    #saving results as json
    save_data_filename=flyer_name[:-4]+"_Ridge_Data.json"
    with open(save_data_filename, "w") as file:
        json.dump(concised_results, file, indent=4)


```
